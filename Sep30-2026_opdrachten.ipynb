# Opdracht 8 - Dict & json

Heel veel API's maken gebruik van json.

- Bij de verwerking in Python moet je json kunnen omzetten in dict formaat en vice versa.
- Dict formaat is relevant voor de interne structuren van Python.
- json formaat is belangrijk voor het verzenden van data over het internet.

**Tip**: [json lint](https://jsonlint.com/)

```python
import json

if_dict = {"ietf-interfaces:interfaces":
           {"interface": [
               {"name": "GigabitEthernet1", "description": "VBox", "type": "iana-if-type:ethernetCsmacd", "enabled": True,
                "ietf-ip:ipv4": {"address": [{"ip": "192.168.56.101", "netmask": "255.255.255.0"}]},
                "ietf-ip:ipv6": {}
               },
               {"name": "Loopback9", "description": "999", "type": "iana-if-type:softwareLoopback", "enabled": True,
                "ietf-ip:ipv4": {"address": [{"ip": "10.9.9.9", "netmask": "255.255.255.0"},
                                             {"ip": "172.29.0.9", "netmask": "255.255.255.0"}]},
                "ietf-ip:ipv6": {}
               }
           ]
           }
          }
print(type(if_dict))
print(if_dict)
if_json = json.dumps(if_dict, indent=2)
print(if_json)
print(type(if_json))
```

```python

terug = json.loads(if_json)
print(type(terug), "| gelijk aan origineel:", terug == if_dict)

for interface in terug["ietf-interfaces:interfaces"]["interface"]:
    ips = [a["ip"] for a in interface["ietf-ip:ipv4"]["address"]]
    print(interface["name"], "->", ips)

# Opslaan en inlezen via een bestand
with open("interfaces.json", "w", encoding="utf-8") as f:
    json.dump(if_dict, f, indent=2)
with open("interfaces.json", "r", encoding="utf-8") as f:
    print("Bestand gelijk aan dict:", json.load(f) == if_dict)
```

# Opdracht 10 - System Information

- In dit opleidingsonderdeel staan systemen en netwerken centraal.
- Het is van belang scripts te beheersen voor het opvragen en managen van systeeminformatie.

```python
!pip install psutil
```

```python

import json
import psutil
import platform
import socket

import datetime
now = datetime.datetime.now()
print(now)


def collect_system_info():
    data = {}

    # CPU info
    data["cpu"] = {
        "platform": platform.platform(),
        "processor": platform.processor(),
        "physical_cores": psutil.cpu_count(logical=False),
        "logical_processors": psutil.cpu_count(logical=True),
        "cpu_freq": psutil.cpu_freq()._asdict() if psutil.cpu_freq() else None,
        "cpu_percent_total": psutil.cpu_percent(interval=0.1),
        "cpu_percent_per_cpu": psutil.cpu_percent(interval=0.1, percpu=True),
    }

    # Memory info
    data["memory"] = psutil.virtual_memory()._asdict()
    data["swap"] = psutil.swap_memory()._asdict()

    # Storage info (alle partities)
    storage = {}
    for part in psutil.disk_partitions(all=False):
        try:
            usage = psutil.disk_usage(part.mountpoint)
            storage[part.mountpoint] = {
                "device": part.device,
                "fstype": part.fstype,
                "opts": part.opts,
                "total": usage.total,
                "used": usage.used,
                "free": usage.free,
                "percent": usage.percent,
            }
        except PermissionError:
            continue
    data["storage"] = storage

    # Network info
    net = {}
    addrs = psutil.net_if_addrs()
    stats = psutil.net_if_stats()
    io = psutil.net_io_counters(pernic=True)
    for iface, addr_list in addrs.items():
        net[iface] = {
            "addrs": [a._asdict() for a in addr_list],
            "stats": stats[iface]._asdict() if iface in stats else None,
            "io": io[iface]._asdict() if iface in io else None,
        }

    data["network"] = {
        "hostname": socket.gethostname(),
        "fqdn": socket.getfqdn(),
        "interfaces": net,
    }

    return data


if __name__ == "__main__":
    info = collect_system_info()
    # JSON output
    print(json.dumps(info, indent=2))
```
